<h2 style="color:DodgerBlue">Название проекта:</h2>

----

### Вариант задания № 3

**Сотрудники компании (Employee)** — иерархия классов сотрудников, демонстрирующая принципы наследования и полиморфизма.

<h2 style="color:DodgerBlue">Описание проекта:</h2>

----

Необходимо создать базовый класс `Employee`, представляющий информацию о сотрудниках компании, и на его основе разработать производные классы, демонстрирующие принципы **наследования** и **полиморфизма**. В каждом из классов реализуются новые атрибуты и методы, а также переопределяются некоторые методы базового класса.

<h3 style="color:DodgerBlue">Базовый класс Employee</h3>

**Атрибуты:**

* `Name` — имя сотрудника;
* `Age` — возраст;
* `Salary` — зарплата.

**Методы:**

* `GetInfo()` — возвращает информацию о сотруднике в виде строки;
* `Work()` — выводит сообщение о выполнении рабочих обязанностей;
* `TakeVacation()` — выводит сообщение о взятии отпуска.

<h3 style="color:DodgerBlue">Производные классы</h3>

| Класс | Дополнительный атрибут | Переопределяемый метод |
| :--- | :--- | :--- |
| `Manager` (Менеджер) | `SubordinatesCount` — количество подчиненных | `Work()` — добавляется информация об управлении командой |
| `Developer` (Разработчик) | `Specialization` — специализация | `TakeVacation()` — добавляется информация о необходимости согласования отпуска с проектными сроками |
| `Designer` (Дизайнер) | `DesignStyle` — стиль дизайна | `Work()` — добавляется информация о работе над дизайн-проектами |

Метод `GetInfo()` переопределяется во всех трех производных классах: через вызов `base.GetInfo()` к общей информации о сотруднике добавляются его должность и собственные атрибуты.

<h3 style="color:DodgerBlue">Используемые принципы ООП</h3>

**Наследование** реализуется записью `class Manager : Employee` — производный класс получает все атрибуты и методы базового класса и дополняет их своими.

**Полиморфизм** реализуется с помощью ключевых слов `virtual` (метод базового класса) и `override` (метод производного класса). Благодаря этому при обращении к объекту через переменную базового типа `Employee` вызывается та реализация метода, которая соответствует фактическому типу объекта.

**Инкапсуляция** реализуется через автоматические свойства (`get`/`set`), которые управляют доступом к данным сотрудника.

<h2 style="color:DodgerBlue">Реализация:</h2>

----

<h4 style="color:DodgerBlue">Базовый класс Employee и производные классы Manager, Developer, Designer:</h4>

```
using System;

public class Employee
{
    public string Name { get; set; }
    public int Age { get; set; }
    public double Salary { get; set; }

    public Employee(string name, int age, double salary)
    {
        Name = name;
        Age = age;
        Salary = salary;
    }

    public virtual string GetInfo()
    {
        return $"Имя: {Name}, Возраст: {Age}, Зарплата: {Salary} руб.";
    }

    public virtual void Work()
    {
        Console.WriteLine($"{Name} выполняет свои рабочие обязанности...");
    }

    public virtual void TakeVacation()
    {
        Console.WriteLine($"{Name} уходит в отпуск...");
    }
}

public class Manager : Employee
{
    public int SubordinatesCount { get; set; }

    public Manager(string name, int age, double salary, int subordinatesCount)
        : base(name, age, salary)
    {
        SubordinatesCount = subordinatesCount;
    }

    public override string GetInfo()
    {
        return base.GetInfo() + $", Должность: Менеджер, Подчиненных: {SubordinatesCount}";
    }

    public override void Work()
    {
        base.Work();
        Console.WriteLine($"{Name} управляет командой из {SubordinatesCount} чел. и распределяет задачи...");
    }
}

public class Developer : Employee
{
    public string Specialization { get; set; }

    public Developer(string name, int age, double salary, string specialization)
        : base(name, age, salary)
    {
        Specialization = specialization;
    }

    public override string GetInfo()
    {
        return base.GetInfo() + $", Должность: Разработчик, Специализация: {Specialization}";
    }

    public override void TakeVacation()
    {
        base.TakeVacation();
        Console.WriteLine($"Отпуск необходимо согласовать с проектными сроками ({Specialization})...");
    }
}

public class Designer : Employee
{
    public string DesignStyle { get; set; }

    public Designer(string name, int age, double salary, string designStyle)
        : base(name, age, salary)
    {
        DesignStyle = designStyle;
    }

    public override string GetInfo()
    {
        return base.GetInfo() + $", Должность: Дизайнер, Стиль дизайна: {DesignStyle}";
    }

    public override void Work()
    {
        base.Work();
        Console.WriteLine($"{Name} работает над дизайн-проектами в стиле \"{DesignStyle}\"...");
    }
}
```

<h4 style="color:DodgerBlue">Демонстрация работы (полиморфизм):</h4>

----

Все созданные объекты помещаются в массив базового типа `Employee[]`. При переборе массива для каждого элемента вызываются одни и те же методы `GetInfo()`, `Work()` и `TakeVacation()`, но выполняется реализация того класса, к которому объект относится на самом деле.

```
Employee employee = new Employee("Иван Петров", 35, 60000);
Manager manager = new Manager("Андрей Смирнов", 42, 120000, 7);
Developer developer = new Developer("Игорь Андреев", 25, 150000, "Backend-разработка");
Designer designer = new Designer("Мария Кузнецова", 29, 95000, "Минимализм");

Employee[] staff = { employee, manager, developer, designer };

foreach (Employee person in staff)
{
    Console.WriteLine(person.GetInfo());
    person.Work();
    person.TakeVacation();
    Console.WriteLine(new string('-', 70));
}
```
