<h2 style="color:DodgerBlue">Название проекта:</h2>

----

### Вариант задания № 3

**Сотрудники компании (Employee)** - иерархия классов сотрудников, демонстрирующая принципы наследования и полиморфизма.

<h2 style="color:DodgerBlue">Описание проекта:</h2>

----

Необходимо создать базовый класс Employee, представляющий информацию о сотрудниках компании, и на его основе разработать производные классы, демонстрирующие принципы наследования и полиморфизма. В каждом из классов реализуются новые атрибуты и методы, а также переопределяются некоторые методы базового класса.

<h3 style="color:DodgerBlue">Базовый класс Employee</h3>

**Атрибуты:**
`Name` - имя сотрудника;
`Age` - возраст;
`Salary` - зарплата.

**Методы:**
`GetInfo()` - возвращает информацию о сотруднике в виде строки;
`Work()` - выводит сообщение о выполнении рабочих обязанностей;
`TakeVacation()` - выводит сообщение о взятии отпуска.

**Производные классы**

Класс | Дополнительный атрибут | Переопределяемый метод

`Manager` (Менеджер) | `SubordinatesCount` - количество подчиненных | `Work()` - добавляется информация об управлении командой

`Developer` (Разработчик) | `Specialization` - специализация | `TakeVacation()` - добавляется информация о необходимости согласования отпуска с проектными сроками

`Designer` (Дизайнер) | `DesignStyle` - стиль дизайна | `Work()` - добавляется информация о работе над дизайн-проектами


<h2 style="color:DodgerBlue">Реализация:</h2>

----

<h4 style="color:DodgerBlue">Базовый класс Employee и производные классы Manager, Developer, Designer:</h4>

```
public class Employee
{
    public string Name { get; set; }
    public int Age { get; set; }
    public int Salary { get; set; }

    public virtual string GetInfo()
    {
        return $"{Name}, {Age} лет, зарплата {Salary} руб.";
    }

    public virtual void Work()
    {
        Console.WriteLine($"{Name} выполняет рабочие обязанности.");
    }

    public virtual void TakeVacation()
    {
        Console.WriteLine($"{Name} уходит в отпуск.");
    }
}

// Менеджер
public class Manager : Employee
{
    public int SubordinatesCount { get; set; }

    public void HoldMeeting()
    {
        Console.WriteLine($"{Name} проводит совещание.");
    }

    public override void Work()
    {
        base.Work();
        Console.WriteLine($"Управляет командой из {SubordinatesCount} человек.");
    }
}

// Разработчик
public class Developer : Employee
{
    public string Specialization { get; set; }

    public void WriteCode()
    {
        Console.WriteLine($"{Name} пишет код ({Specialization}).");
    }

    public override void TakeVacation()
    {
        base.TakeVacation();
        Console.WriteLine("Отпуск необходимо согласовать с проектными сроками.");
    }
}

// Дизайнер
public class Designer : Employee
{
    public string DesignStyle { get; set; }

    public void DrawMockup()
    {
        Console.WriteLine($"{Name} рисует макет.");
    }

    public override void Work()
    {
        base.Work();
        Console.WriteLine($"Работает над дизайн-проектами в стиле {DesignStyle}.");
    }
}

Employee employee = new Employee { Name = "Иван Иванов", Age = 35, Salary = 60000 };
Manager manager = new Manager { Name = "Андрей Мирный", Age = 42, Salary = 120000, SubordinatesCount = 7 };
Developer developer = new Developer { Name = "Арсений Календарёв", Age = 25, Salary = 150000, Specialization = "Backend" };
Designer designer = new Designer { Name = "Дмитрий Кучинский", Age = 29, Salary = 95000, DesignStyle = "минимализм" };

Console.WriteLine(employee.GetInfo());
employee.Work();
employee.TakeVacation();
Console.WriteLine();

Console.WriteLine(manager.GetInfo());
manager.Work();
manager.TakeVacation();
Console.WriteLine();

Console.WriteLine(developer.GetInfo());
developer.Work();
developer.TakeVacation();
Console.WriteLine();

Console.WriteLine(designer.GetInfo());
designer.Work();
designer.TakeVacation();
Console.WriteLine();

Console.WriteLine("Собственные методы производных классов:");
manager.HoldMeeting();
developer.WriteCode();
designer.DrawMockup();
```
